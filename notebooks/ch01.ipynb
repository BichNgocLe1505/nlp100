## 03. 円周率
"Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics."という文を単語に分解し、各単語の（アルファベットの）文字数を先頭から出現順に並べたリストを作成せよ。

Cho câu:

Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics.

Hãy:

tách câu thành các từ,
đếm số chữ cái của từng từ,
tạo một danh sách chứa các số đó theo đúng thứ tự xuất hiện.

```python
sentence = "Now I need a drink, alcoholic of course, after the heavy lectures involving quantum mechanics."
```

```python
# version 1
words = sentence.split(" ")
print(words)

result = []
for word in words:
    result.append(len(word.strip(".,")))
print(result)
```

```python
# version 2
import re
words = sentence.split(" ")
print(words)

result = []
for word in words:
    word = re.sub(r"[^a-zA-Z]", "", word)
    result.append(len(word))
print(result)
```

```python
# version 3
import nltk
from nltk.tokenize import word_tokenize
import string 

words = word_tokenize(sentence)
print(words)

result = []
for word in words:
    if (word not in string.punctuation):
        result.append(len(word))
print(result)
```

```python
# version 4
import nltk
from nltk.tokenize import word_tokenize
import string 

words = word_tokenize(sentence)
print(words)

result = [len(w) for w in words if w not in string.punctuation]
print(result)
```

## 04. 元素記号
"Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."という文を単語に分解し、1, 5, 6, 7, 8, 9, 15, 16, 19番目の単語は先頭の1文字、それ以外の単語は先頭の2文字を取り出し、取り出した文字列から単語の位置（先頭から何番目の単語か）への連想配列（辞書型もしくはマップ型）を作成せよ。

Cho câu:

Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can.

Hãy:

- tách câu thành các từ,
- với các từ ở vị trí:
1, 5, 6, 7, 8, 9, 15, 16, 19
thì lấy 1 ký tự đầu,
- với các từ còn lại thì lấy 2 ký tự đầu,
- tạo một associative array / dictionary / map:
key = chuỗi lấy được,
value = vị trí của từ trong câu.

```python
sentence = "Hi He Lied Because Boron Could Not Oxidize Fluorine. New Nations Might Also Sign Peace Security Clause. Arthur King Can."
```

```python
# version 1
import nltk
from nltk.tokenize import word_tokenize
import string

words = word_tokenize(sentence)
words = [word for word in words if word not in string.punctuation]
print(words)

special = {0, 4, 5, 6, 7, 8, 14, 15, 18}

d = {}

for i, word in enumerate(words):
    if i in special:
        symbol = word[0]
    else:
        symbol = word[:2]
    d[symbol] = i

print(d)
```

```python
# version 2
import nltk
from nltk.tokenize import word_tokenize
import string

words = word_tokenize(sentence)
words = [word for word in words if word not in string.punctuation]
print(words)

special = {0, 4, 5, 6, 7, 8, 14, 15, 18}

d = {}

for i, word in enumerate(words):
    d[word[0] if i in special else word[:2]] = i

print(d)

d = {
    word[0] if i in special else word[:2] : i
    for i, word in enumerate(words)
}

print(d)
```

```python
# version 3
import nltk
from nltk.tokenize import word_tokenize
import string
from collections import defaultdict

words = word_tokenize(sentence)
words = [word for word in words if word not in string.punctuation]
print(words)

special = {0, 4, 5, 6, 7, 8, 14, 15, 18}

d = defaultdict(list)

for i, word in enumerate(words):
    symbol = word[0] if i in special else word[:2]
    d[symbol].append(i)
d
```

## 05. n-gram
与えられたシーケンス（文字列やリストなど）からn-gramを作る関数を作成せよ。この関数を用い、"I am an NLPer"という文から文字tri-gram、単語bi-gramを得よ。

Hãy viết hàm tạo n-gram từ một sequence (chuỗi hoặc list).

Dùng hàm đó để tạo:

character tri-gram,
word bi-gram

từ câu:

I am an NLPer

```python
def build_bi_gram(sequence):
    result = []

    for i in range(len(sequence) - 1):
        result.append((sequence[i], sequence[i + 1]))

    return result
```

```python
sentence = "I am an NLPer"

words = sentence.split()

word_bi_gram = build_bi_gram(words)

print(word_bi_gram)
```

```python
text = "I am an NLPer"

char_bi_gram = build_bi_gram(text)

print(char_bi_gram)
```

```python
def build_tri_gram(sequence):
    result = []

    for i in range(len(sequence) - 2):
        result.append((sequence[i], sequence[i + 1], sequence[i + 2]))
    
    return result
```

```python
sentence = "I am an NLPer"

words = sentence.split()

word_tri_gram = build_tri_gram(words)

print(word_tri_gram)
```

```python
text = "I am an NLPer"

char_tri_gram = build_tri_gram(text)

print(char_tri_gram)
```

```python
# version 1
def build_n_gram(sequence, n):
    result = []

    for i in range(len(sequence) - n + 1):
        result.append(tuple(sequence[i:i+n]))

    return result
```

```python
text = "I am an NLPer"

char_tri_gram = build_n_gram(text, 3)

print(char_tri_gram)
```

```python
sentence = "I am an NLPer"

words = sentence.split()

word_bi_gram = build_n_gram(words, 2)

print(word_bi_gram)
```

## 07. テンプレートによる文生成
引数x, y, zを受け取り「x時のyはz」という文字列を返す関数を実装せよ。さらに、x=12, y="気温", z=22.4として、実行結果を確認せよ。

Hãy cài đặt hàm nhận ba tham số:
- x
- y
- z

và trả về chuỗi:
- x時のyはz

nghĩa là:
- y lúc x giờ là z

Sau đó kiểm tra với:
- x = 12
- y = "気温" (nhiệt độ)
- z = 22.4

```python
# version 1
def string_v1(x, y, z):
    return str(x) + "時の" + y + "は" + str(z)

print(string_v1(12, "気温", 22.4))
```

```python
# version 2
def string_v2(x, y, z):
    return f"{x}時の{y}は{z}"

print(string_v2(12, "気温", 22.4))
```

## 08. 暗号文
与えられた文字列の各文字を、以下の仕様で変換する関数cipherを実装せよ。

* 英小文字ならば (219 - 文字コード) のASCIIコードに対応する文字に置換
* その他の文字はそのまま出力

この関数を用い、英語のメッセージを暗号化・復号化せよ。

Hãy cài đặt hàm cipher biến đổi từng ký tự của chuỗi theo quy tắc:

- nếu là chữ cái thường tiếng Anh:

thay bằng ký tự có mã ASCII:

219−c

(trong đó c là mã ASCII của ký tự gốc)

- các ký tự khác giữ nguyên.

Sau đó:

- mã hóa,
- rồi giải mã

một thông điệp tiếng Anh bằng hàm này.

```python
# version 1
def cipher_v1(sequence):
    result = ""
    
    for char in  sequence:
        if 'a' <= char <= 'z':
            char = chr(219 - ord(char))

        result += char
    return result
```

```python
text = "Hello World!"

encrypted = cipher_v1(text)
print(encrypted)

decrypted = cipher_v1(encrypted)
print(decrypted)
```

```python
# version 2
def cipher_v2(sequence):
    return "".join(transform(x) for x in sequence)

def transform(char):
    if 'a' <= char <= 'z':
        return chr(219 - ord(char))
    else:
        return char
```

```python
text = "Hello World!"

encrypted = cipher_v2(text)
print(encrypted)

decrypted = cipher_v2(encrypted)
print(decrypted)
```

## 09. Typoglycemia
スペースで区切られた単語列に対して、各単語の先頭と末尾の文字は残し、それ以外の文字の順序をランダムに並び替えるプログラムを作成せよ。ただし、長さが４以下の単語は並び替えないこととする。適当な英語の文（例えば"I couldn't believe that I could actually understand what I was reading : the phenomenal power of the human mind ."）を与え、その実行結果を確認せよ。

Hãy viết chương trình xử lý chuỗi gồm các từ cách nhau bởi dấu cách như sau:

- giữ nguyên:
  - chữ cái đầu,
  - chữ cái cuối
- các ký tự ở giữa thì xáo trộn ngẫu nhiên thứ tự.

Tuy nhiên:

- nếu từ có độ dài ≤ 4 thì không xáo trộn.

Hãy thử với câu tiếng Anh thích hợp, ví dụ:

I couldn’t believe that I could actually understand what I was reading : the phenomenal power of the human mind .

và kiểm tra kết quả thực thi.

```python
# version 1
import nltk
from nltk import word_tokenize
import random

sentence = "I couldn’t believe that I could actually understand what I was reading : the phenomenal power of the human mind ."

words = word_tokenize(sentence)
words

result = []
for word in words:
    if len(word) <= 4:
        result.append(word)
        continue

    first = word[0]
    middle = word[1:- 1]
    end = word[- 1]

    # random
    list_of_middle = list(middle)
    random.shuffle(list_of_middle)
    shuffled_middle = "".join(list_of_middle)

    word = first + shuffled_middle + end
    result.append(word)

print(" ".join(result))
```

```python
# version 2 
import nltk
from nltk import word_tokenize
import random
import string

def shuffle_word(word):
    if len(word) <= 4:
        return word

    start = word[0]
    middle = word[1:-1]
    end = word[-1]

    list_of_middle = list(middle)
    random.shuffle(list_of_middle)
    shuffled_middle = "".join(list_of_middle)

    return start + shuffled_middle + end



sentence = "I couldn’t believe that I could actually understand what I was reading : the phenomenal power of the human mind ."

words = word_tokenize(sentence)
result = " ".join(shuffle_word(word) for word in words)
print(result)
```

```python
# version 3
import string

def shuffle_word(word):
    if word in string.punctuation:
        return word
    
    if len(word) <= 4:
        return word

    start = word[0]
    middle = word[1:-1]
    end = word[-1]

    list_of_middle = list(middle)
    random.shuffle(list_of_middle)
    shuffled_middle = "".join(list_of_middle)

    return start + shuffled_middle + end

sentence = "I couldn’t believe that I could actually understand what I was reading : the phenomenal power of the human mind ."
words = sentence.split(" ")
result = " ".join(shuffle_word(word) for word in words)
print(result)
```
