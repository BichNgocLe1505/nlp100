````python
import re
import time

correct = 0
total = 0

for i, row in df.head(20).iterrows():
    prompt = f"""
    {row.iloc[0]}

    A. {row.iloc[1]}
    B. {row.iloc[2]}
    C. {row.iloc[3]}
    D. {row.iloc[4]}

    回答は A, B, C, D のいずれか1文字のみを出力してください。
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    match = re.search(r"\b([ABCD])\b", response.text)

    if match is None:
      continue

    total += 1
    pred = match.group()

    #print("Pred:", pred)
    #print("Gold:", row.iloc[5])
    #print()

    if pred == row.iloc[5]:
      correct += 1

    time.sleep(5)


accuracy = correct / total

print(accuracy)
````

## 43. Độ thiên lệch trong câu trả lời

Dựa trên bài 42, hãy kiểm tra xem độ chính xác có thay đổi khi thay đổi các điều kiện thí nghiệm hay không.

Ví dụ:

- temperature
- prompt
- thứ tự các đáp án
- ký hiệu đáp án (A/B/C/D)

Ví dụ kiểm tra:

- đổi toàn bộ đáp án đúng thành lựa chọn D xem kết quả có thay đổi không.

````python
# version 1 - change temperature

import re
import time
from google.genai import types

correct = 0
total = 0

for i, row in df.head(20).iterrows():
    prompt = f"""
    {row.iloc[0]}

    A. {row.iloc[1]}
    B. {row.iloc[2]}
    C. {row.iloc[3]}
    D. {row.iloc[4]}

    回答は A, B, C, D のいずれか1文字のみを出力してください。
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0.1
        )
    )

    match = re.search(r"\b([ABCD])\b", response.text)

    if match is None:
      continue

    total += 1
    pred = match.group()

    #print("Pred:", pred)
    #print("Gold:", row.iloc[5])
    #print()

    if pred == row.iloc[5]:
      correct += 1

    time.sleep(5)


temp_01_accuracy = correct / total

print(temp_01_accuracy)
````

````python
# version 2 - permute choice

import re
import time
from google.genai import types

correct = 0
total = 0

# A -> C
# B -> A
# C -> D
# D -> B


for i, row in df.head(20).iterrows():
    prompt = f"""
    {row.iloc[0]}

    A. {row.iloc[3]}
    B. {row.iloc[1]}
    C. {row.iloc[4]}
    D. {row.iloc[2]}

    回答は A, B, C, D のいずれか1文字のみを出力してください。
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    match = re.search(r"\b([ABCD])\b", response.text)

    if match is None:
      continue

    total += 1
    pred = match.group()

    #print("Pred:", pred)
    #print("Gold:", row.iloc[5])
    #print()

    mapping = {
    "A": "B",
    "B": "D",
    "C": "A",
    "D": "C"
    }

    if pred == mapping[row.iloc[5]]:
        correct += 1

    time.sleep(5)


permuted_choice_accuracy = correct / total

print(permuted_choice_accuracy)
````

````python
# version 3 - correct answer at D

import re
import time
from google.genai import types

correct = 0
total = 0

# A -> C
# B -> A
# C -> D
# D -> B


for i, row in df.head(20).iterrows():
    choices = [
        row.iloc[1],
        row.iloc[2],
        row.iloc[3],
        row.iloc[4]
    ]

    answer_idx = ord(row.iloc[5]) - ord("A")
    correct_choice = choices[answer_idx]

    wrong_choices = [
        c for i, c in enumerate(choices)
        if i != answer_idx
    ]

    new_choices = wrong_choices + [correct_choice]

    prompt = f"""
    {row.iloc[0]}

    A. {new_choices[0]}
    B. {new_choices[1]}
    C. {new_choices[2]}
    D. {new_choices[3]}

    回答は A, B, C, D のいずれか1文字のみを出力してください。
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    match = re.search(r"\b([ABCD])\b", response.text)

    if match is None:
      continue

    total += 1
    pred = match.group()

    #print("Pred:", pred)
    #print("Gold:", row.iloc[5])
    #print()

    if pred == "D":
        correct += 1

    time.sleep(5)


correct_answer_at_D_accuracy = correct / total

print(correct_answer_at_D_accuracy)
````

## 48. Độ bền vững của đánh giá LLM

Hãy khảo sát mức độ ổn định (hoặc dễ bị thao túng) của hệ thống đánh giá ở bài 47.

Ví dụ:

- đánh giá nhiều lần rồi đo phương sai điểm số
- thêm một câu đặc biệt vào cuối bài thơ để xem điểm có bị thao túng không

````python
# version 1
import re
import time

prompt = f"""
あなたは川柳の審査員です。

次の10句の川柳について、
面白さを10点満点で評価してください。

評価基準:
- ユーモア
- 意外性
- 表現の巧みさ

出力形式:
1: <score>
2: <score>
3: <score>
4: <score>
5: <score>
6: <score>
7: <score>
8: <score>
9: <score>
10: <score>

理由は不要です。

{senryu_text}
"""

all_scores = []

for trial in range(5):

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    scores = []

    for line in response.text.splitlines():

        match = re.match(r"^\d+\s*:\s*(\d+)", line)

        if match:
            scores.append(int(match.group(1)))

    if len(scores) == 10:
        all_scores.append(scores)

    print(f"Trial {trial + 1}")
    print(scores)

    time.sleep(5)
````

````python
import pandas as pd

df_scores = pd.DataFrame(
    all_scores,
    columns=[f"Poem{i}" for i in range(1, 11)]
)

print(df_scores)

print(df_scores.std())
````
